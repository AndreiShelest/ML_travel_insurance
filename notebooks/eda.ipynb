### Categorical Encoding

```python
oh_cols = catalog.load('params:travel_insurance.encoding.one_hot_cols')

X_train = pd.get_dummies(X_train, columns=oh_cols)
```

```python
catboost_cols = catalog.load('params:travel_insurance.encoding.catboost_cols')

CBE_encoder = CatBoostEncoder(cols=catboost_cols)
train_cbe = CBE_encoder.fit_transform(X_train.sort_values(by='net_sales'), y_train)
train_cbe.head(10)
```
